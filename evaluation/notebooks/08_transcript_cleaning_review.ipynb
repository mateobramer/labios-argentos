# Transcript Cleaning Review

Revision semantica/operativa de `transcript_cleaned_stronger`. No entrena, no usa VM, no llama LLM.

```python
import pandas as pd
from IPython.display import Markdown, display

from evaluation.src.transcript_cleaning_review import (
    cargar_outputs,
    cambios_por_tipo,
    candidatos_no_autoaplicados,
    decision,
    ejemplos_autoaplicados,
    impacto_train,
    resumen,
    top_fuentes_bad,
    top_fuentes_cambios,
    usability_counts,
)

pd.set_option("display.max_colwidth", 160)
changes, candidates, policy = cargar_outputs()
```

## Resumen

```python
display(resumen(changes, candidates, policy))
display(usability_counts(policy))
```

## Cambios autoaplicados

```python
display(cambios_por_tipo(changes))
display(ejemplos_autoaplicados(changes, 10))
```

## Candidatos no autoaplicados

```python
if candidates.empty:
    display(Markdown("No hay candidatos agresivos pendientes."))
else:
    display(candidates.groupby("candidate_type").size().rename("clips").reset_index().sort_values("clips", ascending=False))
    display(candidatos_no_autoaplicados(candidates, 30))
```

## Bad candidates / review

```python
display(policy[policy["transcript_usability"].isin(["questionable", "bad_candidate"])].head(30))
display(top_fuentes_bad(policy))
```

## Ejemplos por fuente

```python
display(top_fuentes_cambios(changes))
if not candidates.empty:
    display(candidates.groupby("source_id").size().rename("candidates").reset_index().sort_values("candidates", ascending=False).head(15))
```

## Impacto esperado en train

```python
display(impacto_train(policy))
```

## Decision: listo / revisar / bloquear

```python
display(Markdown(decision(changes, candidates, policy)))
```
