Revision visual de `lower_face_resized96`. Funciona aunque los ROIs variant full no existan todavia.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, Markdown, display

from evaluation.src.preprocessing_variant_review import (
    cargar_smoke,
    pending_message,
    preview_groups,
    preview_paths,
    sample_table,
    smoke_decision,
    smoke_status,
)

pd.set_option("display.max_colwidth", 160)
smoke = cargar_smoke()
status = smoke_status(smoke)
decision = smoke_decision(smoke)
```

## Estado del smoke

## Contact sheets / previews

```python
previews = preview_paths()
groups = preview_groups()
if previews.empty:
    display(Markdown("Pendiente de VM: falta generar lower_face_resized96."))
else:
    display(groups.head(20))
    for _, row in groups.head(3).iterrows():
        display(Markdown(f"### {row['sample']}"))
        for label, column in [
            ("Current ROI 96x96", "current_roi_96x96"),
            ("lower_face_resized96", "lower_face_resized96"),
            ("Side-by-side", "side_by_side"),
        ]:
            path = row.get(column)
            if isinstance(path, str) and path:
                display(Markdown(f"**{label}**"))
                display(Image(filename=path))
```
